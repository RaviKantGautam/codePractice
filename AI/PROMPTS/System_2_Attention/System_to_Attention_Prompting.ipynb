```python
import ollama
```

```python
prompt="""The number of each Arts Campus' T&T Supermarket equals 3. The number of each Engineering Campus' T&T Supermarket equals 4.
The number of each Science Park's Zion Market equals 1 more than each Arts Campus' T&T Supermarket.
The number of each Arts Campus' Seafood Supermarket equals the sum of each Science Park's Zion Market, Arts Campus' T&T Supermarket and each Arts Campus'
Meat Market. The number of each Arts Campus' Meat Market equals 4 times as much as each Science Park's Zion Market. How many Meat Market does
Arts Campus have?"""
```

```python
final_prompt = f"""**Role:** You are a strict System 2 Attention (S2A) Context Extractor. Your objective is to de-noise unstructured inputs while preserving 100% of the essential data needed to fulfill the request.

**Processing Rules:**
1. **Identify the Core Intent:** Determine the exact question the user is asking or the specific task they want executed.
2. **Extract Dependencies (CRITICAL):** Identify all vital context provided in the text. This includes rules, factual premises, numerical values, code snippets, constraints, and required formats. You MUST NOT delete any information that dictates how the core task should be completed.
3. **Scrub Only True Noise:** Delete ONLY conversational filler (e.g., greetings, pleasantries), emotional language, or completely irrelevant background tangents.
4. **Reconstruct:** Rebuild the prompt into a clean, structured format. Present the essential context/dependencies first, followed by the direct, actionable core instruction.

**Strict Output Constraints:**
- Do NOT answer the user's question or execute the core task.
- Do NOT include any introductory or concluding text (e.g., no "Here is the prompt:").
- Output ONLY the reconstructed, optimized prompt.

**Input Query:**
{prompt}
"""
```

```python
final_prompt
```

```python
output = ollama.generate(
    model='gpt-oss:120b-cloud',
    prompt=final_prompt
)

op= output["response"]
print(output["response"])
```

```python
output = ollama.generate(
    model='llama3',
    prompt=op,
    options={'temperature': 0.1}
)

print(output["response"])
```
