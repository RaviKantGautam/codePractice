```python
import ollama
```

```python
topic= "Do world need AI"
```

#### GETTING RELEVANT TOPICS AND FACTS ABOUT THE TOPIC

```python
research_prompt = f"""Topic: {topic}

Output ONLY a bulleted list of sub-topics and hard facts.
NO introductions, NO conclusions, NO explanations, NO filler words.


Sub-topics:
-

Facts:
-
"""
```

```python
research_prompt
```

```python
output = ollama.generate(
    model='gpt-oss:120b-cloud',
    prompt=research_prompt
)

op= output["response"]
print(output["response"])
```

#### WRITING BLOG USING THE INFORMATION RECEIVED.

```python
# The writer prompt formatted as a multi-line f-string
writer_prompt = f"""You are an expert technical writer. Your task is to write a comprehensive, professional blog post based
entirely on the provided topic and the research blueprint.

Topic: {topic}

Research Blueprint:
{op}
"""
```

```python
writer_prompt
```

```python
output = ollama.generate(
    model="llama3",
    prompt=writer_prompt
)

blog= output["response"]
print(output["response"])
```

#### CRITIC PROMPT

```python
# Assuming 'blog_post' holds the full text from your writer prompt
critic_prompt = f"""Review this blog post for technical depth, engagement, and structure.

Blog Post Content:
{blog}

Output ONLY a bulleted list of specific improvements(if any).
NO conversational filler. NO introductions. NO conclusions.

Areas for Improvement:
-
"""
```

```python
output = ollama.generate(
    model='gpt-oss:120b-cloud',
    prompt=critic_prompt
)

critic= output["response"]
print(output["response"])
```

#### FINAL DRAFT

```python
# Assuming 'original_blog' and 'critic_feedback' are variables from your previous steps
refiner_prompt = f"""Task: Rewrite the blog post by incorporating the specific improvements suggested by the critic.

Original Blog:
{blog}

Critic Feedback:
{critic}

Strict Instructions:
1. Apply every point from the Critic Feedback to the Original Blog.
2. Maintain a professional, technical tone.
3. Output ONLY the final, polished blog post.
4. NO introductions, NO "Here is the revised version," and NO commentary.
"""
```

```python
output = ollama.generate(
    model="llama3",
    prompt=refiner_prompt
)

blog= output["response"]
print(output["response"])
```
