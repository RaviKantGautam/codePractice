## GENERAL WAY OF CALLING MODEL

```python
import ollama

# The System Instruction defined above
system_instruction = (
    "You are a helpful and precise AI assistant. "
    "1. First, provide your full explanation and reasoning. "
    "2. CRITICAL: Your response must end with a strictly separate new line. "
    "3. On this last line, print ONLY the final value in bold (e.g., **144**). "
    "4. Do NOT include labels, equations (like 'x ='), or any other text on this last line."
)
prompt = "Solve this: 15 + 10 * 2"

output = ollama.generate(
    model='llama3',
    prompt=prompt,
    system=system_instruction,
    options={'temperature': 0.1}
)

print(output["response"])
```

## SELF CONSISTENT PROMPTING

```python
from collections import Counter
import re
```

```python
#collections is a built-in Python module that provides specialized container data types — alternatives to Python’s general-purpose ones like list, dict,
#set, and tuple.

#Counter is a class inside collections used for counting how many times items appear.
```

```python
# SELF CONSISTENCY PROMPTING.

answers = []

for i in range(5):
    output = ollama.generate(
        model='llama3',
        prompt=prompt,
        system=system_instruction,
        options={'temperature': 0.6}
    )

    text = output['response'].strip()

    # Simple regex to find content inside ** **
    match = re.search(r'\*\*(.*)\*\*', text)

    if match:
        result = match.group(1) # .group(1) is the text inside the bold marks
        answers.append(result)
        print(f"Attempt {i+1}: {result}")
```

```python
answers
```

```python
# Find the most frequent answer
final_answer = Counter(answers).most_common(1)[0][0]

print(f"\n🏆 Final Consensus: {final_answer}")
```

## UNIVERSAL SELF CONSISTENT PROMPTING.

```python
import ollama

# The open-ended question
question = "Which planet in our solar system is the largest?"


candidate_responses = []

for i in range(3):
    # Generating multiple versions locally
    output = ollama.generate(
        model='llama3',
        prompt=question,
        options={'temperature': 0.8}
    )
    candidate_responses.append(output['response'])

print(candidate_responses)
```

```python
for i in candidate_responses:
    print(i)
    print("**************")
```

```python
# Build the prompt for the "Judge" model manually
evaluation_prompt = f"I have generated these 3 recommendations for: {question}\n\n"

# Use a for loop to label each response for the judge
for index, response in enumerate(candidate_responses):
    evaluation_prompt += f"Response {index}: {response}\n\n"

evaluation_prompt
```

```python
# Adding a strict instruction to get a clean result
evaluation_prompt += (
    "Select the most consistent response based on majority consensus."
    "Provide ONLY the text of the selected response. Do not explain your choice."
)

evaluation_prompt
```

```python
# 3. FINAL SELECTION
# Use a more powerful model to act as the judge [cite: 11]
final_decision = ollama.generate(
    model='gpt-oss:120b-cloud',
    prompt=evaluation_prompt
)

print("--- SELECTED CONSISTENT RESPONSE ---")
print(final_decision['response'])
```

```python
# Real-world scenario: Summarizing a customer meeting transcript
patient_history = """
Patient: Elias Thorne, Age 58

History of Present Illness
The patient is a 58-year-old male who presented to the clinic complaining of persistent, dull aching in his lower lumbar region that has radiated down his left leg over the past three weeks. He describes the pain as a 6/10 on the visual analog scale, which intensifies after prolonged sitting or during early morning hours. He denies any recent trauma but notes that the symptoms began shortly after he started a more intensive gardening routine. Over-the-counter ibuprofen has provided only transient relief, and he reports a new onset of "pins and needles" sensations in his left foot.

Past Medical History
Mr. Thorne has a ten-year history of well-controlled Type 2 Diabetes Mellitus, managed with Metformin and lifestyle modifications. He also has a history of mild hypertension and was diagnosed with seasonal asthma in his early thirties. Surgical history is notable for a successful laparoscopic appendectomy in 2012 and a minor procedure for a meniscus tear in his right knee five years ago. He maintains an active lifestyle but has recently struggled with weight management due to decreased mobility from his current back pain.

Social and Family History
The patient is a non-smoker who works as a freelance graphic designer, which involves significant desk time. He lives with his spouse and two dogs and enjoys light hiking and woodworking. His family history is significant for cardiovascular disease, with his father suffering a myocardial infarction at age 62. His mother is alive and well at 84 with a history of osteoporosis. There are no known drug allergies, though he reports a mild sensitivity to certain laundry detergents.
"""

print(patient_history)
```

```python
import ollama


generation_instruction = (
    "You are an expert medical scribe. "
    "Analyze the following patient history and generate a structured Clinical Abstract. "
    "The abstract must strictly include: \n"
    "1. Demographics & Chief Complaint\n"
    "2. Key Symptoms & History of Present Illness\n"
    "3. Relevant Medical History\n"
    "4. Lifestyle/Social Factors\n\n"
    "Do not hallucinate information not present in the text.\n"
    f"Patient Data:\n{patient_history}"
)

summary_candidates = []
for i in range(3):
    response = ollama.generate(
        model='llama3',
        prompt=generation_instruction,
        options={'temperature': 0.6}
    )
    summary_candidates.append(response['response'])


judge_prompt = (
    "You are a Senior Medical Editor. I will present three different summaries "
    "generated from the same patient history.\n"
    "Your task is to select the BEST summary based on the following criteria:\n"
    "1. Accuracy: Does it accurately reflect the patient's symptoms (dull ache, radiation, pins and needles)?\n"
    "2. Completeness: Does it capture the diabetes history and the new gardening routine?\n"
    "3. Consistency: Is it consistent with the facts without adding hallucinated details?\n\n"
)

for idx, summary in enumerate(summary_candidates):
    judge_prompt += f"Summary {idx}: {summary}\n\n"

judge_prompt += (
    "Evaluate the candidates. Return ONLY the text of the single best summary. "
    "Do not provide explanations, just the summary text."
)

```

```python
print(judge_prompt)
```

```python
# Using a larger "Cloud" or powerful model for the selection
final_consensus = ollama.generate(
    model='gpt-oss:120b-cloud',
    prompt=judge_prompt
)

print("--- SELECTED CLINICAL SUMMARY ---")
print(final_consensus['response'])
```
